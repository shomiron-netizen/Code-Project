```python
import cv2
```

```python
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/images.jpeg')
```

```python
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
reflacted_x = cv2.flip(image, 0)
```

```python
reflacted_y = cv2.flip(image, 1)
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,3,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(reflacted_x)
plt.title('Reflected about X-axis')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(reflacted_y)
plt.title('Reflected about Y-axis')
plt.axis('off')
plt.show()
```
