```python
import cv2
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/images.jpeg')
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
h, w = image.shape[:2]
```

```python
tx = 100
ty = 50
```

```python
M = np.float32([
    [1,0,tx],
    [0,1,ty]
])
```

```python
translate = cv2.warpAffine(image, M, (w,h))
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(translate)
plt.title('Translated Image')
plt.axis("off")
plt.show()
```
